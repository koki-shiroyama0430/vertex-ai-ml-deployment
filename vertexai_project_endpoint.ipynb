```python
from google.cloud import aiplatform

# ID
PROJECT_ID = 'project-bfba9093-76af-4977-a3a'
ENDPOINT_ID = '6754096519776829440'
LOCATION = 'us-central1'

# Vertex AI 初期化
aiplatform.init(project=PROJECT_ID, location=LOCATION)

# エンドポイントに接続
endpoint = aiplatform.Endpoint(ENDPOINT_ID)

# Dummy Data
# All Black
test_instance = [0.0] * 784

print('クラウドへ推論リクエスト送信中…')

# 推論実行
try:
  response = endpoint.predict(instances=[test_instance])
  print('\n 推論成功！')
  print('AIの予測結果（確率）：')
  # 0~9の確立を表示するループ
  for i, prob in enumerate(response.predictions[0]):
    print(f'数字{i}: {prob:.4f}')

# 最も確率の高い数字を表示
  import numpy as np
  prediction = np.argmax(response.predictions[0])
  print(f'\n →AIは、この数字を「{prediction}」だと判定しました！')

except Exception as e:
  print('\n ✖ エラーが発生しました。認証が通っていないか、IDが間違っている可能性があります。')
  print(e)
```
